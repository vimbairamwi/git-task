```python
#Import libraries

import pandas as pd
import spacy
from spacytextblob.spacytextblob import SpacyTextBlob

# Load the spaCy medium English model
nlp = spacy.load("en_core_web_md")

# Add TextBlob pipeline for sentiment analysis
nlp.add_pipe("spacytextblob")
```

```python
import pandas as pd

df = pd.read_csv(
    "Datafiniti_Amazon_Consumer_Reviews_of_Amazon_Products_May19.csv"
)

df.head()
```

```python
#Data cleaning
# Select reviews column

reviews_data = df["reviews.text"]

# Remove rows with missing review text
clean_data = df.dropna(subset=["reviews.text"])

print("Number of reviews:", len(clean_data))
```

```python
#Reprocessing function
def preprocess_review(review):
    """
    Cleans review text by:
    1. Converting to lowercase
    2. Removing leading/trailing spaces
    3. Removing stop words
    4. Removing punctuation
    """

    review = str(review).lower().strip()

    doc = nlp(review)

    cleaned_tokens = [
        token.text
        for token in doc
        if not token.is_stop
        and not token.is_punct
        and not token.is_space
    ]

    return " ".join(cleaned_tokens)
```

```python
#Apply preprocessing
# Create new column containing cleaned reviews

clean_data["cleaned_review"] = (
    clean_data["reviews.text"]
    .apply(preprocess_review)
)

clean_data[["reviews.text", "cleaned_review"]].head()
```

```python
#Sentiment analysis
def sentiment_analysis(review):
    """
    Determines sentiment of a product review.

    Parameters:
        review (str): Product review text

    Returns:
        sentiment_label (str)
        polarity_score (float)
    """

    doc = nlp(review)

    polarity = doc._.blob.polarity

    if polarity > 0:
        sentiment = "Positive"

    elif polarity < 0:
        sentiment = "Negative"

    else:
        sentiment = "Neutral"

    return sentiment, polarity
```

```python
#Test mdel
sample_reviews = [
    clean_data["reviews.text"].iloc[0],
    clean_data["reviews.text"].iloc[50],
    clean_data["reviews.text"].iloc[100]
]

for review in sample_reviews:

    sentiment, polarity = sentiment_analysis(review)

    print("=" * 60)
    print("Review:")
    print(review)
    print()
    print("Sentiment:", sentiment)
    print("Polarity Score:", polarity)
```

```python
#Compare similarity
# Select two reviews

review_1 = clean_data["reviews.text"].iloc[0]
review_2 = clean_data["reviews.text"].iloc[1]

doc1 = nlp(review_1)
doc2 = nlp(review_2)

similarity_score = doc1.similarity(doc2)

print("Review 1:")
print(review_1)

print("\nReview 2:")
print(review_2)

print("\nSimilarity Score:")
print(similarity_score)
```

```python
#Analyse a larger sampleof reviews
# Analyse first 100 reviews

results = clean_data.head(100).copy()

results["sentiment"] = results["cleaned_review"].apply(
    lambda x: sentiment_analysis(x)[0]
)

results["polarity"] = results["cleaned_review"].apply(
    lambda x: sentiment_analysis(x)[1]
)

results[["reviews.text", "sentiment", "polarity"]].head()
```
