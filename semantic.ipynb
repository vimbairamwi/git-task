```python
#load dataset
import spacy

# Load the medium-sized English model
nlp = spacy.load('en_core_web_md')
```

```python
#Comparing
word1 = nlp("cat")
word2 = nlp("monkey")
word3 = nlp("banana")

# Print the similarity scores
print(word1.similarity(word2))
print(word3.similarity(word2))
print(word3.similarity(word1))
```

The similarity score between cat and monkey is relatively high because both are animals. Banana and monkey also show noticeable similarity because monkeys are commonly associated with bananas. Cat and banana have the lowest similarity because they belong to different categories and have little semantic relationship.


```python
#Example of self
word1 = nlp("car")
word2 = nlp("bus")
word3 = nlp("apple")

print(word1.similarity(word2))
print(word3.similarity(word2))
print(word3.similarity(word1))
```

The words car and bus have a higher similarity score because both are vehicles used for transportation. Apple has a lower similarity score with both words because it is a fruit and belongs to a completely different category.

```python
# Process the sentence to tokenise the words
tokens = nlp('cat apple monkey banana')

# Iterate over each pair of tokens
for token1 in tokens:
    for token2 in tokens:
        # Print the text and similarity score
        print(token1.text, token2.text, token1.similarity(token2))
```

Words compared with themselves always return 1.0 because they are identical. Animal-related words and fruit-related words tend to have higher similarity scores because they share semantic meanings.

```python
sentence_to_compare = "Why is my cat on the car"

sentences = [
    "where did my dog go",
    "Hello, there is my car",
    "I've lost my car in my car",
    "I'd like my boat back",
    "I will name my dog Diana"
]

model_sentence = nlp(sentence_to_compare)

for sentence in sentences:
    similarity = nlp(sentence).similarity(model_sentence)
    print(sentence + " - " + str(similarity))
```

Sentences containing words related to cat, dog, or car generally receive higher similarity scores because they share context with the original sentence.

```python
import spacy

nlp = spacy.load("en_core_web_sm")

word1 = nlp("cat")
word2 = nlp("monkey")
word3 = nlp("banana")

print(word1.similarity(word2))
print(word3.similarity(word2))
print(word3.similarity(word1))
```

The en_core_web_md model generally produces more meaningful and accurate similarity scores because it contains larger word vectors. The en_core_web_sm model is smaller and may give less reliable similarity results. The medium model better captures relationships between words and sentences, making it more suitable for semantic similarity tasks.
